```python
from wordcloud import WordCloud, STOPWORDS
import matplotlib.pyplot as plt 

texto = 'Estar em toda a parte onde houver bem a fazer e sofrimento a aliviar'

#texto = 'PDF significa Portable Document Format (Formato de Documento Portátil).'\'É um formato de arquivo criado pela Adobe em 1993 para partilhar documentos sem perder a formatação original.'

# Conjunto de palavras que vão ser removidas do texto 

as_minhas_stopwords = set(['em' , 'a' , 'toda', 'onde' , 'e'])

# ----------------------------------Configurar e gerir a nuvem a partir do texto----------------------------------------

nuvem = WordCloud (
    width = 800,
    height = 400,
    background_color = "#DDEEE7",
    colormap='GnBu', # outas cores no classroom

    # 'plasma', 'Blues', 'viridis', 'Greens', 'cool', 'winter', 'YlGnBu', 'GnBu', 'BuGn', 'ocean', 'flag'
    max_words= 7,
    stopwords=as_minhas_stopwords  # remover as minhas palavras
).generate(texto)

# ----------------Exibir a imagem a partir da Matplotlib (biblioteca python, usada para criar gráficos)----------------


plt.figure(figsize=(10,5)) #tamanho da exibição da figura 
plt.imshow(nuvem, interpolation= 'bilinear') # renderiza a imagem
plt.axis('off')
plt.show()

nuvem.to_file('nuvem_palavras.png')
```
