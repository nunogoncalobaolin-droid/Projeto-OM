## 2. Ler a média

Aceita vírgula ou ponto (ex: `16,4` ou `16.4`) e só aceita valores entre 10 e 20.

```python
def pedir_media():
    """Pede a média até o aluno escrever um número válido entre 10 e 20."""
    while True:
        texto = input("A tua média esperada (10 a 20 valores): ").strip().replace(",", ".")
        try:
            media = float(texto)
        except ValueError:
            print("Isso não é um número. Exemplo: 16.2")
            continue
        if 10 <= media <= 20:
            return media
        print("Introduz uma média válida entre 10 e 20 valores.")
```

## 3. Escolher o curso

```python
def pedir_curso():
    print("\nCursos disponíveis:")
    for i, curso in enumerate(NOMES_CURSOS, start=1):
        print(f"   {i}) {curso}")
    while True:
        escolha = input("Número do curso pretendido: ").strip()
        if escolha.isdigit() and 1 <= int(escolha) <= len(NOMES_CURSOS):
            return NOMES_CURSOS[int(escolha) - 1]
        print(f"Escolhe um número entre 1 e {len(NOMES_CURSOS)}.")
```

## 4. Escala visual

Desenha uma linha de 10 a 20 com a média de referência (`|`) e a média do aluno (`●`).

```python
def barra(media_aluno, media_ref, largura=30):
    """Desenha uma escala de 10 a 20 com a média de referência (|) e a do aluno (●)."""
    linha = ["─"] * (largura + 1)
    pos_ref = round((media_ref - 10) / 10 * largura)
    pos_aluno = round((media_aluno - 10) / 10 * largura)
    linha[pos_ref] = "|"
    linha[pos_aluno] = "●"
    return "10 " + "".join(linha) + " 20"
```

## 5. Simular a candidatura

Ordena as universidades da média mais alta para a mais baixa, conta em quantas a média chega e mostra o resultado de cada uma, com as notas dos 5 anos.

```python
def simular(media, curso):
    universidades = sorted(BASE_DADOS_UNIVERSIDADES.get(curso, []), key=lambda u: u["mediaUltimos5Anos"], reverse=True)
    if not universidades:
        print("De momento não existem dados para este curso.")
        return

    conseguidas = sum(1 for u in universidades if media >= u["mediaUltimos5Anos"])
    print(f"\n📊 {conseguidas} de {len(universidades)} universidades em {curso} "
          f"têm média de referência igual ou inferior a {media:.1f}\n")

    for uni in universidades:
        entra = media >= uni["mediaUltimos5Anos"]
        estado = "✓ Média suficiente" if entra else "✕ Abaixo da média"
        anos = " · ".join(f"{2021 + i}: {nota:.1f}" for i, nota in enumerate(uni["notas"]))
        print(f"{uni['nome']} — {uni['curso']}")
        print(f"   Média do último colocado (2021–2025): {uni['mediaUltimos5Anos']:.1f}  →  {estado}")
        print(f"   {anos}")
        print(f"   {barra(media, uni['mediaUltimos5Anos'])}")
        print(f"   Ver na DGES: {uni['link']}\n")
```

## 6. Programa principal

```python
def executar():
    print("\n📊  SIMULADOR DE MÉDIAS E UNIVERSIDADES")
    while True:
        media = pedir_media()
        curso = pedir_curso()
        simular(media, curso)
        if input("Fazer outra simulação? (s/n): ").strip().lower() != "s":
            break
```
