## 2. Bibliotecas e grupos de dificuldade

Os 5 níveis de dificuldade da base de dados estão agrupados em 3 grupos.

```python
import unicodedata
from urllib.parse import quote_plus, urlparse

GRUPOS = {
    "dificil":  ("Difícil",  "Médias muito altas, testes de admissão ou entrevistas e poucas vagas."),
    "moderado": ("Moderado", "Média boa e requisitos de língua, sem processos muito seletivos."),
    "facil":    ("Fácil",    "Acesso aberto ou só com o secundário concluído e a língua exigida."),
}
```

## 3. Funções auxiliares

`carregar_dados` devolve a base de dados; `escolher_numero` lê um número válido (ou **V** para voltar); `link_curso` cria o link de pesquisa que leva à página do curso no site da universidade.

```python
def carregar_dados():
    return DADOS_EUROPA

def escolher_numero(pergunta, maximo, permitir_voltar=True):
    """Pede um número entre 1 e maximo. Devolve None se o aluno escrever V (voltar)."""
    while True:
        texto = input(pergunta).strip().upper()
        if permitir_voltar and texto == "V":
            return None
        if texto.isdigit() and 1 <= int(texto) <= maximo:
            return int(texto) - 1
        print(f"Escreve um número entre 1 e {maximo}" + (" ou V para voltar." if permitir_voltar else "."))

def link_curso(universidade, curso):
    """Link de pesquisa que leva à página do curso no site da universidade."""
    dominio = urlparse(universidade["site"]).netloc.replace("www.", "")
    return "https://www.google.com/search?q=" + quote_plus(f"site:{dominio} {curso['nome']}")
```

## 4. Passo 1 — Escolher o país

Mostra os países por ordem alfabética, com o número de universidades de cada grupo.

```python
# ---------------------------------------------------------------- PASSO 1
def escolher_pais(dados):
    # ordena ignorando acentos (para "Áustria" ficar no A)
    paises = sorted(dados.keys(), key=lambda p: unicodedata.normalize("NFD", p).encode("ascii", "ignore").decode())
    print("\n🌍  PASSO 1 — ESCOLHE UM PAÍS")
    for i, pais in enumerate(paises, start=1):
        unis = dados[pais]
        contagem = " · ".join(
            f"{sum(1 for u in unis if u['dificuldade'] == g)} {GRUPOS[g][0].lower()}"
            for g in GRUPOS if any(u["dificuldade"] == g for u in unis))
        print(f"{i:>3}) {pais:<22} {len(unis):>2} univ.  ({contagem})")
    indice = escolher_numero("\nNúmero do país (ou V para sair): ", len(paises))
    return None if indice is None else paises[indice]
```

## 5. Passo 2 — Universidades por dificuldade

```python
# ---------------------------------------------------------------- PASSO 2
def escolher_universidade(pais, universidades, so_ingles):
    print(f"\n🎓  PASSO 2 — UNIVERSIDADES EM {pais.upper()}")

    lista = []
    for chave, (nome_grupo, descricao) in GRUPOS.items():
        do_grupo = [u for u in universidades if u["dificuldade"] == chave
                    and (not so_ingles or any("Inglês" in c["linguas"] for c in u["cursos"]))]
        if not do_grupo:
            continue
        print(f"\n● {nome_grupo.upper()} ({len(do_grupo)}) — {descricao}")
        for uni in sorted(do_grupo, key=lambda u: u["nome"]):
            lista.append(uni)
            em_ingles = sum(1 for c in uni["cursos"] if "Inglês" in c["linguas"])
            print(f"{len(lista):>4}) {uni['nome']} — {uni['cidade']}  "
                  f"[{len(uni['cursos'])} curso{'s' if len(uni['cursos']) != 1 else ''}, {em_ingles} em inglês]")

    if not lista:
        print("Nenhuma universidade corresponde ao filtro.")
        return None
    indice = escolher_numero("\nNúmero da universidade (ou V para voltar): ", len(lista))
    return None if indice is None else lista[indice]
```

## 6. Passo 3 — Cursos da universidade

```python
# ---------------------------------------------------------------- PASSO 3
def mostrar_universidade(pais, uni):
    nome_grupo = GRUPOS[uni["dificuldade"]][0]
    print("\n" + "=" * 70)
    print(f"{uni['nome']}  —  {uni['cidade']}, {pais}")
    print("=" * 70)
    print(f"Dificuldade de entrada: {nome_grupo}")
    print(f"Requisitos (referência): {uni['requisitos']}")
    print(f"Sobre: {uni['descricao']}")
    print(f"Site oficial: {uni['site']}")
    print("\nLicenciaturas abertas a alunos internacionais:")
    for i, curso in enumerate(uni["cursos"], start=1):
        print(f"  {i}. {curso['nome']}  ({', '.join(curso['linguas'])})")
        print(f"     Ver curso: {link_curso(uni, curso)}")
    print("\nℹ️  Valores de referência — confirma sempre no site oficial.")
    input("\nCarrega em Enter para voltar às universidades...")
```

## 7. Programa principal

```python
def executar():
    dados = carregar_dados()
    total = sum(len(l) for l in dados.values())
    print(f"\n🌍  ESTUDAR NA EUROPA — {len(dados)} países, {total} universidades")

    while True:
        pais = escolher_pais(dados)
        if pais is None:
            break
        so_ingles = input("Mostrar só universidades com cursos em inglês? (s/n): ").strip().lower() == "s"
        while True:
            uni = escolher_universidade(pais, dados[pais], so_ingles)
            if uni is None:
                break
            mostrar_universidade(pais, uni)
```
