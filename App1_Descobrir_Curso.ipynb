## 3. Mostrar uma pergunta e ler a resposta

Mostra a barra de progresso e as opções A–D. Se o aluno escrever **V**, volta à pergunta anterior. Respostas inválidas são recusadas.

```python
def pedir_resposta(pergunta, numero, total):
    """Mostra uma pergunta e devolve o valor da opção escolhida (ou None para voltar atrás)."""
    print(f"\nPergunta {numero} de {total}  [{'█' * numero}{'░' * (total - numero)}] {round(numero / total * 100)}%")
    print(pergunta["texto"])
    for i, opcao in enumerate(pergunta["opcoes"]):
        print(f"   {LETRAS[i]}) {opcao['texto']}")

    while True:
        dica = " (ou V para voltar)" if numero > 1 else ""
        escolha = input(f"A tua resposta [A-D]{dica}: ").strip().upper()
        if escolha == "V" and numero > 1:
            return None
        if escolha in LETRAS[:len(pergunta["opcoes"])] and escolha != "":
            return pergunta["opcoes"][LETRAS.index(escolha)]["valor"]
        print("Opção inválida. Escreve A, B, C ou D.")
```

## 4. Calcular o resultado

Regras de pontuação:
- cada resposta dá **2 pontos** ao curso correspondente;
- a resposta *“prefiro pessoas a cálculos”* (`humanas`) dá **1 ponto** a Direito, Recursos Humanos e Marketing;
- ganha o curso com mais pontos.

```python
def calcular_resultado(respostas):
    """Soma os pontos de cada curso e devolve a chave do curso vencedor."""
    pontuacao = {curso: 0 for curso in DADOS_CURSOS}
    for valor in respostas:
        if valor == "humanas":
            pontuacao["direito"] += 1
            pontuacao["rh"] += 1
            pontuacao["marketing"] += 1
        elif valor in pontuacao:
            pontuacao[valor] += 2

    # Em caso de empate ganha o primeiro da lista (igual à versão web)
    vencedor = max(pontuacao, key=pontuacao.get)
    return vencedor, pontuacao
```

## 5. Mostrar o resultado

```python
def mostrar_resultado(chave):
    info = DADOS_CURSOS[chave]
    print("\n" + "=" * 60)
    print("O CURSO MAIS INDICADO PARA TI É:")
    print(f"   >>> {info['nome'].upper()} <<<")
    print("=" * 60)
    print(info["descricao"])
    print("\nOutros cursos que também te podem interessar:")
    for relacionado in info["relacionados"]:
        print(f"   • {relacionado}")
```

## 6. Programa principal

Junta tudo: faz as 8 perguntas, calcula e mostra o resultado, e pergunta se o aluno quer refazer o teste.

```python
def executar():
    print("\n🧭  DESCOBRE O TEU CURSO IDEAL")
    print("Responde a 8 perguntas sobre os teus interesses.")

    while True:
        respostas = [None] * len(PERGUNTAS)
        indice = 0
        while indice < len(PERGUNTAS):
            valor = pedir_resposta(PERGUNTAS[indice], indice + 1, len(PERGUNTAS))
            if valor is None:          # voltar à pergunta anterior
                indice -= 1
                continue
            respostas[indice] = valor
            indice += 1

        vencedor, _ = calcular_resultado(respostas)
        mostrar_resultado(vencedor)

        if input("\nQueres refazer o questionário? (s/n): ").strip().lower() != "s":
            break
```
